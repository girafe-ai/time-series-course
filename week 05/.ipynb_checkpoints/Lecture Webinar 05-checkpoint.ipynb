# Lecture

## Definitions

$$
    X_t = R \cos (\omega t + \phi) + Z_t
$$

- (Angular) frequency $\omega$
- (Cycle) frequency $f = \omega / 2 \pi$
- Period or wavelength $T = 1/f = 2 \pi / \omega$
- Amplitude $R$
- Phase $\phi$

## Spectral representation

$$
    X_t = R \cos (\omega t + \phi) + Z_t
$$

$$
    X_t = \sum_{j=1}^k R_j \cos (\omega_j t + \phi_j) + Z_t
$$
with $\mathbb{E}[R_j] = 0$ or $\phi_j \sim U[0, 2\pi]$

$$
    X_t = \sum_{j=1}^k \left( a_j \cos \omega_j t + b_j \sin \omega_j t \right) + Z_t
$$

$$
    X_t = \int_0^\pi \cos \omega t {\rm d} u(\omega) + \int_0^\pi \sin \omega t {\rm d} v(\omega)
$$

## (Power) Spectral Distribution Function

Wiener–Khinchin theorem (spectral representation of ACVF):
$$
    \gamma(k) = \int_0^\pi \cos \omega k {\rm d} F(\omega)
$$

Properties:
- Monotonically increasing
- $F(0) = 0$
- $F(\pi) = \mathbb{V}\text{ar}(X_t) = \sigma_X^2$

## Notes on spectral distribution function

1. Integral limits from $0$ to $\pi$, because for discrete-time process measured at unit intervals of time:
   $$
    \cos{(\omega + k \pi) t} = \begin{cases}
        \cos{\omega t}, \; k \text{ even}, \\
        \cos{( \pi - \omega) t}, \; k \text{ odd}
    \end{cases}
   $$

   Frequency $\omega = \pi$ is called Nyquist frequency.

2. Interpretation: Spectral Distribution Function $F(\omega)$ is the contribution to the variance of the series, which is accounted for by frequencies in the range $(0, \omega)$.

3. If the process contains a deterministic component like $X = R \cos\left(\omega_0 t + \phi\right)$, where $R$ is constant and $\phi \sim U[0, 2\pi]$, then $F(\omega)$ is a step function with a step at $\omega_0$ of height
   $$
    \mathbb{V}\text{ar}(X) = \mathbb{E}\left[R^2 \cos^2 \left( \omega_0 t + \phi \right)\right] = \frac12 R^2
   $$

4. Define
   $$
    F^\ast(\omega) = F(\omega) / \sigma^2_X
   $$

## (Power) Spectral Density Function = Spectrum

$$
    f(\omega) = F'(\omega)
$$

$$
    \gamma(k) = \int_0^\pi \cos \omega k f(\omega) {\rm d} \omega
$$

$$
    f(\omega) = \frac1\pi \sum_{k=-\infty}^\infty \gamma(k) e^{- i\omega k} = \frac1\pi \left( \gamma(0) + 2 \sum_{k=1}^\infty \gamma(k) \cos{\omega k} \right)
$$

## Notes on spectral density function

1. If the process is stationary, the integral will actually not converge. The theorem still holds, but shouldn't be written like that.

2. If we allow frequencies to be negative, we can change their range as follows:
    $$
        \gamma(k) = \int_{-\pi}^\pi e^{i \omega k} f(\omega) {\rm d} \omega
    $$
    
    $$
        f(\omega) = \frac{1}{2\pi} \sum_{k=-\infty}^\infty \gamma(k) e^{- i\omega k}
    $$

3. If we normalize similarly to spectral distribution function, ACF will appear:
$$
    f^\ast(\omega) = f(\omega) / \sigma_X^2 = \frac1\pi \left( 1 + 2 \sum_{k=1}^\infty \rho(k) \cos{\omega k} \right)
$$

4. For a continuous purely indeterministic stationary process $X(t)$:
    $$
        f(\omega) = \frac1\pi \int_{-\infty}^\infty \gamma(\tau) e^{-i\omega \tau} {\rm d} \tau = \frac2\pi \int_0^\infty \gamma(\tau) e^{-i\omega \tau} {\rm d} \tau
    $$

    $$
        \gamma(\tau) = \int_0^\infty f(\omega) \cos{\omega \tau} {\rm d} \omega
    $$

## Example 1: spectrum of white noise

For white noise $Z_t$ with $\mathbb{V}\text{ar}(Z_t) = \sigma^2_Z$, using
$$
    f(\omega) = \frac1\pi \left( \gamma(0) + 2 \sum_{k=1}^\infty \gamma(k) \cos{\omega k} \right) = \gamma(0) / \pi = \sigma^2_Z / \pi
$$

## Example 2: spectrum of MA(1)

Consider $X_t = Z_t + \beta Z_{t-1}$. We know its ACF:
$$
    \rho(k) = \begin{cases}
        1, \; k = 0, \\
        \beta / (1 + \beta^2), \; k = \pm 1, \\
        0, \; \text{else}
    \end{cases}
$$

Using
$$
    f^\ast(\omega) = \frac1\pi \left( 1 + 2 \sum_{k=1}^\infty \rho(k) \cos{\omega k} \right) = \frac1\pi \left( 1 + \frac{2 \beta}{1 + \beta^2} \cos{\omega} \right)
$$

Then,
$$
    f(\omega) = \sigma^2_X f^\ast(\omega) = (1 + \beta^2) \sigma^2_Z f^\ast(\omega)
$$

```python
w = np.linspace(0, np.pi, 100)
sigma_z = 1
beta = 1
f_pos = (1 + beta ** 2) * sigma_z ** 2 * (1 + 2 * beta * np.cos(w) / (1 + beta ** 2)) / np.pi
f_neg = (1 + beta ** 2) * sigma_z ** 2 * (1 - 2 * beta * np.cos(w) / (1 + beta ** 2)) / np.pi
```

```python
fig, ax = plt.subplots(figsize=(16,9))
ax.plot(w, f_pos, label="$\\beta = 1$")
ax.plot(w, f_neg, label="$\\beta = -1$")
ax.set_xlabel("frequency, $\\omega$")
ax.set_ylabel("spectrum, $f(\\omega)$")
ax.set_title("Spectrum of MA(1)")
ax.legend();
```

## Example 3: spectrum of AR(1)

Consider $X_t = \alpha X_{t-1} + Z_t$. We know its ACVF $\gamma(k) = \sigma_X^2 \alpha^{|k|}$. Using
$$
    \begin{aligned}
    f(\omega) & = \frac1\pi \sum_{k=-\infty}^\infty \gamma(k) e^{- i\omega k} = \\
    & = \frac{\sigma_X^2}{\pi} \left( 1 + \sum_{k=1}^\infty \alpha^k e^{- i\omega k} + \sum_{k=1}^\infty \alpha^k e^{i\omega k} \right) = \\
    & = \frac{\sigma_X^2}{\pi} \left( 1 + \frac{\alpha e^{-i \omega}}{1 - \alpha e^{-i \omega}} + \frac{\alpha e^{i \omega}}{1 - \alpha e^{i \omega}} \right) = \\
    & = \frac{\sigma_X^2}{\pi} \frac{1 - \alpha^2}{1 - 2 \alpha \cos{\omega} + \alpha^2} = \frac{\sigma^2_Z}{\pi(1 - 2 \alpha \cos{\omega} + \alpha^2)}
    \end{aligned}
$$

```python
w = np.linspace(0, np.pi, 100)
sigma_z = 1
alpha = 0.7
f_pos = sigma_z ** 2 / (np.pi * (1 - 2 * alpha * np.cos(w) + alpha ** 2))
f_neg = sigma_z ** 2 / (np.pi * (1 + 2 * alpha * np.cos(w) + alpha ** 2))
```

```python
fig, ax = plt.subplots(figsize=(16,9))
ax.plot(w, f_pos, label="$\\alpha = 1$")
ax.plot(w, f_neg, label="$\\alpha = -1$")
ax.set_xlabel("frequency, $\\omega$")
ax.set_ylabel("spectrum, $f(\\omega)$")
ax.set_title("Spectrum of AR(1)")
ax.legend();
```

## Example 4: spectrum of AR(2)

Consider $X_t = \alpha_1 X_{t-1} + \alpha_2 X_{t-2} + Z_t$. 

Without derivation,
$$
 f(\omega) = \frac{\sigma^2_Z}{\pi \left( 1 + \alpha_1^2 + \alpha_2^2 - 2 \alpha_1 (1 - \alpha_2) \cos{\omega} - 2 \alpha_2 \cos{2\omega} \right)}
$$

```python
def ar_2_f(w, sigma_z, alpha_1, alpha_2):
    return sigma_z ** 2 / (np.pi * (1 + alpha_1 ** 2 + alpha_2 ** 2 - 2 * alpha_1 * (1 - alpha_2) * np.cos(w) - 2 * alpha_2 * np.cos(2 * w)))
```

```python
w = np.linspace(0, np.pi, 100)
sigma_z = 1
f_left = ar_2_f(w, sigma_z, 0.5, 0.1)
f_mid = ar_2_f(w, sigma_z, 0.5, -0.5)
f_right = ar_2_f(w, sigma_z, -0.5, 0.1)
f_mid_low = ar_2_f(w, sigma_z, 0.1, 0.5)
```

```python
fig, ax = plt.subplots(figsize=(16,9))
ax.plot(w, f_left, label="$\\alpha_1 = 0.5, \\alpha_2 = 0.1$")
ax.plot(w, f_right, label="$\\alpha_1 = -0.5, \\alpha_2 = 0.1$")
ax.plot(w, f_mid, label="$\\alpha_1 = 0.5, \\alpha_2 = -0.5$")
ax.plot(w, f_mid_low, label="$\\alpha_1 = 0.1, \\alpha_2 = 0.5$")
ax.set_xlabel("frequency, $\\omega$")
ax.set_ylabel("spectrum, $f(\\omega)$")
ax.set_title("Spectrum of AR(2)")
ax.legend();
```

## Example 5: spectrum of deterministic process

Consider $X_t = \cos (\omega_0 t + \phi)$, where $\phi \sim U[0, 2\pi]$. As discussed previously,
$$
    \gamma(k) = \frac12 \cos \omega_0 k
$$
$$
    F(\omega) = \begin{cases}
        0, \; \omega < \omega_0, \\
        \frac12, \; \omega \geqslant \omega_0
    \end{cases}
$$
$$
    f(\omega) = 0, \; \forall \omega \neq \omega_0
$$

## Spectral analysis

Spectral analysis is the name given to methods of estimating the spectral density function, or spectrum, of a given time series. Traditional spectral analysis is essentially a modification of Fourier analysis so as to make it suitable for stochastic rather than deterministic functions of time. The Fourier series representation has $N$ parameters to describe $N$ observations and so can be made to fit the data exactly. The overall effect of the Fourier analysis of the data is to partition the variability of the series intocomponents at the specified $N$ frequencies called harmonics.

## Discrete Fourier Transform

Given time series $X_t$, DFT transforms it into the same number of frequency components $\omega_k$:
$$
    \omega_k = \sum_{t=0}^{T-1} X_t e^{- 2 \pi i k t / T}
$$

## Parseval's theorem

The Fourier transform is **unitary**. Loosely, that the sum of the square of a function is equal to the sum of the square of its transform:
$$
    \sum_{t=0}^{T-1} |X_t|^2 = \frac1T \sum_{k=0}^{T-1} |\omega_k|^2 = \sum_{k=0}^{T-1} \frac1T |\omega_k|^2 = \sum_{k=0}^{T-1} I(\omega_k)
$$

## Periodogram

$$
I(\omega) = \frac1T |\omega|^2
$$

Plot $I(\omega)$ vs $\omega$ is called periodogram.

## Properties of periodogram

1. Periodogram is the DFT of ACVF:
   $$
    I(\omega_p) = \frac1\pi \left( c_0 + 2 \sum_{k=1}^{T-1} c_k \cos{\omega_p k} \right)
   $$
3. Asymptotically unbiased estimate of the power spectrum
4. Not consistent estimate!

## Better estimates

$$
\hat{f}(\omega) = \frac1\pi \left( \lambda_0 c_0 + 2 \sum_{k=0}^M \lambda_k c_k \cos{\omega k} \right)
$$

1. Tukey window
   $$
    \lambda_k = \frac12 \left( 1 + \cos{\frac{\pi k}{M}} \right)
   $$

```python
from darts import TimeSeries
import darts.datasets as ds
import scipy.stats as sts
import statsmodels.api as sm
import statsmodels.tsa.api as tsa
from darts.utils.statistics import plot_acf
```

```python
from scipy.signal import periodogram
```

```python
from scipy.signal import find_peaks
```

```python
from astsadata import speech, cmort
```

```python
cmort_ts = TimeSeries.from_series(cmort.set_index(
    pd.date_range(
        cmort.index[0].start_time.date(),
        periods=508,
        freq="W-MON",
        inclusive="both"
    )
))
```

```python
fig, ax = plt.subplots(figsize=(16,9))
cmort_ts.plot(ax=axs)
```

```python
freq, psd = periodogram(cmort_ts.values().flatten(), fs=52.18)
```

```python
fig, ax = plt.subplots(figsize=(16,9))
ax.plot(freq, psd)
ax.set_xlabel("frequency, years")
```

```python
freq[find_peaks(psd, threshold=50)[0]]
```

```python
cmort_sdiff_ts = cmort_ts.diff(52)
```

```python
fig, ax = plt.subplots(figsize=(16,9))
cmort_sdiff_ts.plot(ax=ax)
```

```python
freq2, psd2 = periodogram(cmort_sdiff_ts.values().flatten(), fs=52.18)
```

```python
fig, ax = plt.subplots(figsize=(16,9))
ax.plot(freq2, psd2)
ax.set_xlabel("frequency, years")
```

```python
fig, ax = plt.subplots(figsize=(16,9))
ax.semilogy(52.18/freq2[-100:], psd2[-100:])
ax.set_xlabel("weeks")
```

```python
speech_ts = TimeSeries.from_series(speech.reset_index(drop=True))
```

```python
fig, ax = plt.subplots(figsize=(16,9))
speech_ts.plot(ax=ax)
```

```python
freq3, psd3 = periodogram(speech_ts.values().flatten())
```

```python
fig, ax = plt.subplots(figsize=(16,9))
ax.plot(freq3, psd3)
```
